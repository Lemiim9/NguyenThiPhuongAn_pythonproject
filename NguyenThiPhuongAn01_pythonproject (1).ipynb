```python
# #1. TẠO DATAFRAME CHỨA THÔNG TIN 10 SINH VIÊN
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Ho_ten": [
        "Nguyễn Minh Anh",
        "Trần Gia Huy",
        "Lê Hoàng Yến",
        "Đặng Châu Thuyền",
        "Thái Mỹ Hoa",
        "Phạm Tuấn Kiệt",
        "Bùi Khánh Linh",
        "Phan Đức Long",
        "Đỗ Mai Phương",
        "Hoàng Nhật Nam"],
    "Chuyen_can": [9.0, 7.5, 8.0, 6.0, 6.7, 5.5, 8.5, 8.02, 10.0, 6.5],
    "Giua_ky":    [8.5, 8.0, 7.0, 5.0, 6.7, 6.5, 8.0, 8.03, 10.0, 7.5],
    "Cuoi_ky":    [9.2, 7.0, 8.5, 4.5, 6.7, 7.0, 8.0, 7.08, 10.0, 8.1]}
df = pd.DataFrame(data)
ten_cot = {
    "Ho_ten": "Họ và tên",
    "Chuyen_can": "Điểm chuyên cần",
    "Giua_ky": "Điểm giữa kỳ",
    "Cuoi_ky": "Điểm cuối kỳ",
    "Tong_ket": "Điểm tổng kết",
    "Xep_loai": "Xếp loại"}
display(df.rename(columns=ten_cot))
```

```python

# #2. TÍNH ĐIỂM TỔNG KẾT

df["Tong_ket"] = (
    df["Chuyen_can"] * 0.2
    + df["Giua_ky"] * 0.3
    + df["Cuoi_ky"] * 0.5
).round(2)

# Hiển thị bảng sau khi thêm điểm tổng kết
display(df.rename(columns=ten_cot))
```

```python

# #3. XẾP LOẠI SINH VIÊN

def xep_loai(diem):
    if diem >= 8.5:
        return "Giỏi"
    elif diem >= 7.0:
        return "Khá"
    elif diem >= 5.0:
        return "Trung bình"
    else:
        return "Yếu"

df["Xep_loai"] = df["Tong_ket"].apply(xep_loai)

display(df.rename(columns=ten_cot))
```

```python

# #4. THỐNG KÊ ĐIỂM SINH VIÊN

# 1. Tính điểm tổng kết trung bình của lớp
diem_trung_binh = df["Tong_ket"].mean()

# 2. Tìm sinh viên có điểm tổng kết cao nhất
sv_cao_nhat = df.loc[df["Tong_ket"].idxmax()]

# 3. Tìm sinh viên có điểm tổng kết thấp nhất
sv_thap_nhat = df.loc[df["Tong_ket"].idxmin()]

# 4. Đếm số sinh viên đạt từ 5 điểm trở lên
so_sinh_vien_dat = (df["Tong_ket"] >= 5).sum()

# In kết quả thống kê
print(f"Điểm tổng kết trung bình của lớp: {diem_trung_binh:.2f}")

print(
    f"Sinh viên cao nhất: {sv_cao_nhat['Ho_ten']} "
    f"- {sv_cao_nhat['Tong_ket']:.2f} điểm"
)

print(
    f"Sinh viên thấp nhất: {sv_thap_nhat['Ho_ten']} "
    f"- {sv_thap_nhat['Tong_ket']:.2f} điểm"
)

print(f"Số sinh viên đạt (điểm >= 5): {so_sinh_vien_dat}/10")
```

```python

# #5. TRỰC QUAN HÓA ĐIỂM TỔNG KẾT BẰNG MATPLOTLIB

# Tạo khung biểu đồ
fig, ax = plt.subplots(figsize=(12, 6))

# Vẽ biểu đồ cột
cot = ax.bar(
    df["Ho_ten"],
    df["Tong_ket"]
)

# Thiết lập tiêu đề và tên trục
ax.set_title("ĐIỂM TỔNG KẾT CỦA 10 SINH VIÊN", fontsize=14)
ax.set_xlabel("Họ và tên sinh viên")
ax.set_ylabel("Điểm tổng kết")

# Điểm số có thang từ 0 đến 10
ax.set_ylim(0, 10)

# Hiển thị tên sinh viên nghiêng để dễ đọc
ax.tick_params(axis="x", rotation=40)

for nhan in ax.get_xticklabels():
    nhan.set_ha("right")

# Hiển thị điểm trên đầu mỗi cột
ax.bar_label(cot, fmt="%.2f", padding=3)

# Thêm đường lưới ngang
ax.grid(axis="y", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

# Căn chỉnh và hiển thị biểu đồ
plt.tight_layout()
plt.show()
```
